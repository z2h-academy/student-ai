# 03 — Prompt básico: roles y estructura

**Level 1 — LLM Engineering**

Comparamos el MISMO prompt con y sin `system prompt`:

- **Caso 1**: solo el mensaje del usuario (el modelo improvisa el comportamiento)
- **Caso 2**: con `system` (reglas explícitas: idioma, longitud, tono)

La pregunta es la misma; cambia el *comportamiento*.

```python
import requests

OLLAMA_HOST = "http://localhost:11434"


def chat(messages: list[dict], model: str = "llama3.2") -> str:
    """Envia una conversacion completa a Ollama y devuelve la respuesta."""
    url = f"{OLLAMA_HOST}/api/chat"
    payload = {
        "model": model,
        "messages": messages,
        "stream": False,
    }
    response = requests.post(url, json=payload, timeout=120)
    response.raise_for_status()
    return response.json()["message"]["content"]
```

## Caso 1 — Sin system prompt

```python
pregunta = "Explica que es una API en maximo 2 oraciones."

sin_system = [
    {"role": "user", "content": pregunta},
]

print("=== CASO 1: SIN system prompt ===")
print(chat(sin_system))
```

## Caso 2 — Con system prompt

```python
con_system = [
    {
        "role": "system",
        "content": (
            "Eres un asistente tecnico en espanol. "
            "Respondes siempre en espanol, en maximo 2 oraciones, "
            "con lenguaje simple y sin jerga tecnica."
        ),
    },
    {"role": "user", "content": pregunta},
]

print("=== CASO 2: CON system prompt ===")
print(chat(con_system))
```

## Conclusión

El `system` no aporta conocimiento nuevo: aporta **reglas de comportamiento**. Misma pregunta, respuestas con estructura y tono distintos.
